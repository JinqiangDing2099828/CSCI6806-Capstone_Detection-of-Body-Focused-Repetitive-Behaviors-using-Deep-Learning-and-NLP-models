## Library Imports

This cell imports the necessary libraries for data handling and numerical computation. 
`pandas` and `numpy` will be used for processing the sensor dataset, and `Path` helps manage file directories.


```python
from pathlib import Path
import pandas as pd
import numpy as np


```

## Directory Initialization

This cell sets the project root path and creates folders for raw and processed data. 
It ensures the correct directory structure before loading or saving any files.



```python
ROOT = Path.cwd()
if not (ROOT / "data").exists() and (ROOT.parent / "data").exists():
    ROOT = ROOT.parent

RAW_DIR = ROOT / "data" / "raw"
PROCESSED_DIR = ROOT / "data" / "processed"

RAW_DIR.mkdir(parents=True, exist_ok=True)
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

print("Project root:", ROOT)
```

## Data Loading

This cell reads the raw training data and displays the total number of rows and unique sequences. 
This helps confirm that the dataset is loaded correctly.


```python
train_path = RAW_DIR / "train.csv"
raw_df = pd.read_csv(train_path)

print("Raw rows:", raw_df.shape[0])
print("Raw sequences:", raw_df["sequence_id"].nunique())
```

## Missing Value Analysis

This cell checks how many rows contain NULL values and calculates the percentage of affected data.



```python

rows_with_null = raw_df.isnull().any(axis=1)

print("Rows with NULL:", rows_with_null.sum())
print("Percentage removed:", rows_with_null.mean() * 100)
```

## NULL Removal

This cell removes rows with missing values and shows the updated dataset size.


```python
df_no_null = raw_df[~rows_with_null].copy()

print("After NULL removal:")
print("Rows:", df_no_null.shape[0])
print("Sequences:", df_no_null["sequence_id"].nunique())
```

## Sequence Statistics

This cell calculates the average sequence length, its standard deviation, and defines a threshold for filtering short sequences.


```python
seq_lengths = df_no_null.groupby("sequence_id").size()

mean_len = seq_lengths.mean()
std_len = seq_lengths.std()
threshold = mean_len - std_len

mean_len, std_len, threshold
```

## Short Sequence Detection

This cell finds and counts sequences that are shorter than the threshold.


```python
short_sequences = seq_lengths[seq_lengths < threshold]

len(short_sequences), short_sequences.sort_values()
```

## Filtering Short Sequences

This cell removes sequences that are shorter than the threshold and displays the updated dataset size.


```python
df_clean = df_no_null[
    ~df_no_null["sequence_id"].isin(short_sequences.index)
].copy()

print("After short sequence removal:")
print("Rows:", df_clean.shape[0])
print("Sequences:", df_clean["sequence_id"].nunique())
```

## Export Processed Data

This cell saves the cleaned dataset to the processed data folder for use in later stages of the project.



```python
output_path = PROCESSED_DIR / "train_cleaned.csv"
df_clean.to_csv(output_path, index=False)

output_path
```
