```python
# =================================================================
# TOF Preprocessing - Alignment Fix
# 1. Force use of global IDs from 'split_test_seq_ids.npy' 
#    to match Model 2 (1518 samples).
# 2. Added patch for Numpy 2.0 vs 1.0 compatibility issues.
# 3. Output format: (64, 5, 8, 8) for CNN input.
# =================================================================
```

```python
import os
from pathlib import Path
import numpy as np
import pandas as pd


#
# Find route path
#
NOTEBOOK_DIR = Path.cwd()

# model → notebooks
def find_repo_root(start_path: Path):
    cur = start_path
    for _ in range(10):
        if (cur / "data").exists():
            return cur
        cur = cur.parent
    return start_path
REPO_ROOT = find_repo_root(NOTEBOOK_DIR)
RAW_DIR = REPO_ROOT / "data" / "processed"
OUT_DIR = REPO_ROOT / "data" / "processed_data_tof"
OUT_DIR.mkdir(parents=True, exist_ok=True)


PREDS_DIR = REPO_ROOT / "results" / "preds"

parquet_path = RAW_DIR / "final_processed_data.parquet"

print("NOTEBOOK_DIR:", NOTEBOOK_DIR)
print("REPO_ROOT   :", REPO_ROOT)
print("Parquet     :", parquet_path)
print("Exists?     :", parquet_path.exists())

GROUP_COL = "subject"
LABEL_COL = "gesture"
SEQ_COL   = "sequence_id"

#
# Read Parquet
# 
df = pd.read_parquet(parquet_path)
print("Loaded parquet:", df.shape)

for c in [GROUP_COL, LABEL_COL, SEQ_COL]:
    if c not in df.columns:
        raise ValueError(f"Missing column: {c}")

# Read TOF Rows
tof_cols = [c for c in df.columns if str(c).lower().startswith("tof_")]
print("Found TOF cols:", len(tof_cols))
if len(tof_cols) != 320:
    print("Warning: TOF cols not 320, check naming.")

def tof_sort_key(name):
    # "tof_3_v12"
    parts = name.split("_")
    sensor = int(parts[1])
    v = int(parts[2][1:])
    return (sensor, v)

tof_cols_sorted = sorted(tof_cols, key=tof_sort_key)

# 
# Gesture mapping
# 
gesture_classes = sorted(df[LABEL_COL].dropna().unique().tolist())
gesture_to_idx = {g: i for i, g in enumerate(gesture_classes)}
np.save(OUT_DIR / "gesture_classes_raw.npy",
        np.array(gesture_classes, dtype=object),
        allow_pickle=True)
print("Num gesture classes:", len(gesture_classes))

# 
#  BFRB Rules
#
BFRB_KEYWORDS = [
    "Cheek - pinch skin",
    "Forehead - pull hairline",
    "Neck - scratch",
    "Neck - pinch skin",
    "Eyelash - pull hair",
    "Eyebrow - pull hair",
    "Forehead - scratch",
    "Above ear - pull hair",
]

def is_bfrb_gesture(g):
    if not isinstance(g, str):
        return 0
    return 1 if any(k in g for k in BFRB_KEYWORDS) else 0

# 
# 
#    

import sys
if 'numpy._core' not in sys.modules:
    import numpy.core.multiarray
    sys.modules['numpy._core'] = sys.modules['numpy.core']
    sys.modules['numpy._core.multiarray'] = sys.modules['numpy.core.multiarray']
print("Loading global sequence IDs to ensure exact alignment with Model 2...")
seqid_train_raw = np.load(PREDS_DIR / "split_train_seq_ids.npy", allow_pickle=True)
seqid_test_raw  = np.load(PREDS_DIR / "split_test_seq_ids.npy", allow_pickle=True)

seqid_train = np.array([str(x) for x in seqid_train_raw])
seqid_test  = np.array([str(x) for x in seqid_test_raw])

np.save(OUT_DIR / "seqid_train.npy", seqid_train, allow_pickle=True)
np.save(OUT_DIR / "seqid_test.npy",  seqid_test,  allow_pickle=True)
print("Unified split loaded successfully:")
print("  train:", len(seqid_train))
print("  test :", len(seqid_test)) 

# 
#    Output each simple(T_FIXED, 5, 8, 8)
#
T_FIXED = 64

df = df.dropna(subset=tof_cols_sorted + [SEQ_COL, GROUP_COL, LABEL_COL])

train_set = set(seqid_train.tolist())
test_set  = set(seqid_test.tolist())

def pad_truncate(seq_arr, T=T_FIXED):
    Tseq = seq_arr.shape[0]
    if Tseq >= T:
        return seq_arr[:T]
    pad_len = T - Tseq
    pad = np.zeros((pad_len, 5, 8, 8), dtype=np.float32)
    return np.concatenate([seq_arr, pad], axis=0)

def build_X_y_for_seqids(df_rows, wanted_set):
    X_list, y18_list, ybin_list, subj_list, sid_list = [], [], [], [], []

    for sid, g in df_rows.groupby(SEQ_COL, sort=False):
        if sid not in wanted_set:
            continue

        X = g[tof_cols_sorted].to_numpy(dtype=np.float32) 
        X = X.reshape(-1, 5, 8, 8)                        
        X_fixed = pad_truncate(X, T_FIXED)                 

        gesture_str = g[LABEL_COL].iloc[0]
        subject_val = g[GROUP_COL].iloc[0]

        y18 = gesture_to_idx[gesture_str]
        ybin = is_bfrb_gesture(gesture_str)

        X_list.append(X_fixed)
        y18_list.append(y18)
        ybin_list.append(ybin)
        subj_list.append(subject_val)
        sid_list.append(sid)

    X = np.stack(X_list, axis=0)
    y18 = np.array(y18_list, dtype=np.int64)
    ybin = np.array(ybin_list, dtype=np.int64)
    subj = np.array(subj_list)
    sid  = np.array(sid_list, dtype=object)
    return X, y18, ybin, subj, sid

X_train, y18_train, ybin_train, subj_train, sid_train = build_X_y_for_seqids(df, train_set)
X_test,  y18_test,  ybin_test,  subj_test,  sid_test  = build_X_y_for_seqids(df, test_set)

print("TOF train:", X_train.shape, "TOF test:", X_test.shape)

# 
# Saved（CNN-BiLSTM + late fusion）
#
np.save(OUT_DIR / "tof_train.npy", X_train)
np.save(OUT_DIR / "tof_test.npy",  X_test)

np.save(OUT_DIR / "label_train_18.npy", y18_train)
np.save(OUT_DIR / "label_test_18.npy",  y18_test)

np.save(OUT_DIR / "label_train_binary.npy", ybin_train)
np.save(OUT_DIR / "label_test_binary.npy",  ybin_test)

np.save(OUT_DIR / "group_train_subject.npy", subj_train, allow_pickle=True)
np.save(OUT_DIR / "group_test_subject.npy",  subj_test,  allow_pickle=True)

np.save(OUT_DIR / "seqid_train_effective.npy", sid_train, allow_pickle=True)
np.save(OUT_DIR / "seqid_test_effective.npy",  sid_test,  allow_pickle=True)

print("Saved to:", OUT_DIR)
```

```python
### 2. MODEL ARCHITECTURE: CNN-BiLSTM
# - Input Shape: (Batch, Time=64, Channels=5, Height=8, Width=8) [cite: 134]
# - Feature Extraction: A 2D-CNN processes each TOF frame (5 sensors) through 
#   two convolutional layers (32 and 64 kernels)[cite: 136].
# - Sequence Modeling: A Bidirectional LSTM (BiLSTM) captures temporal 
#   dependencies, producing a 256-dimensional embedding[cite: 137].
# - Classifier: A fully connected head maps embeddings to 18 gesture classes[cite: 138].
```

```python
import os
from pathlib import Path
import json, time
import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import f1_score
import matplotlib.pyplot as plt

# 
# Set up root repo
# 
NOTEBOOK_DIR = Path.cwd()

# model → notebooks 
REPO_ROOT = NOTEBOOK_DIR.parent.parent

DATA_DIR  = REPO_ROOT / "data" / "processed_data_tof"
PLOTS_DIR = REPO_ROOT / "plots"
PLOTS_DIR.mkdir(parents=True, exist_ok=True)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("NOTEBOOK_DIR:", NOTEBOOK_DIR)
print("REPO_ROOT   :", REPO_ROOT)
print("Data dir    :", DATA_DIR)
print("Plots dir   :", PLOTS_DIR)
print("Device      :", device)
SEED = 42
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

#
#  Read the TOF sequence input + label
#  Now tof_train.npy should be (N, 64, 5, 8, 8)
# 
TOF_TRAIN_PATH = DATA_DIR / "tof_train.npy"
TOF_TEST_PATH  = DATA_DIR / "tof_test.npy"

tof_train = np.load(TOF_TRAIN_PATH)  # (N,T,5,8,8)
tof_test  = np.load(TOF_TEST_PATH)

label_train_18 = np.load(DATA_DIR / "label_train_18.npy")
label_test_18  = np.load(DATA_DIR / "label_test_18.npy")
label_train_bin = np.load(DATA_DIR / "label_train_binary.npy")
label_test_bin  = np.load(DATA_DIR / "label_test_binary.npy")

gesture_classes = np.load(DATA_DIR / "gesture_classes_raw.npy", allow_pickle=True)

print("tof_train:", tof_train.shape)
print("tof_test :", tof_test.shape)
print("label_train_18 unique:", len(np.unique(label_train_18)))
print("label_test_18  unique:", len(np.unique(label_test_18)))

# BFRB 8 labels
# Determine which are BFRBs from the saved gesture_classes.
BFRB_KEYWORDS = [
    "Cheek - pinch skin",
    "Forehead - pull hairline",
    "Neck - scratch",
    "Neck - pinch skin",
    "Eyelash - pull hair",
    "Eyebrow - pull hair",
    "Forehead - scratch",
    "Above ear - pull hair",
]

def is_bfrb_name(name: str) -> bool:
    if not isinstance(name, str):
        return False
    return any(k in name for k in BFRB_KEYWORDS)

BFRB_CLASS_INDICES = [i for i, g in enumerate(gesture_classes) if is_bfrb_name(g)]
print("BFRB_CLASS_INDICES:", BFRB_CLASS_INDICES)
print("BFRB class names:", [gesture_classes[i] for i in BFRB_CLASS_INDICES])

assert len(BFRB_CLASS_INDICES) == 8, "BFRB classes should be 8. Check gesture names/keywords."

bfrb_map = {orig_idx: new_idx for new_idx, orig_idx in enumerate(BFRB_CLASS_INDICES)}

def convert_18_to_9_array(labels_18):
    labels_9 = np.empty_like(labels_18)
    for i, y in enumerate(labels_18.astype(int)):
        if y in bfrb_map:
            labels_9[i] = bfrb_map[y]
        else:
            labels_9[i] = 8
    return labels_9


# 
# Dataset / DataLoader
# 
class TOFSeqDataset(Dataset):
    def __init__(self, x, y18, ybin):
        # x: (N,T,5,8,8)
        self.x = torch.tensor(x, dtype=torch.float32)
        self.y18 = torch.tensor(y18, dtype=torch.long)
        self.ybin = torch.tensor(ybin, dtype=torch.float32)

    def __len__(self):
        return self.x.shape[0]

    def __getitem__(self, idx):
        return self.x[idx], self.y18[idx], self.ybin[idx]

BATCH_SIZE = 256
train_ds = TOFSeqDataset(tof_train, label_train_18, label_train_bin)
test_ds  = TOFSeqDataset(tof_test,  label_test_18,  label_test_bin)

train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True)
test_loader  = DataLoader(test_ds,  batch_size=BATCH_SIZE, shuffle=False)

print("Train batches:", len(train_loader))
print("Test batches :", len(test_loader))

# 
# D) CNN-BiLSTM Model（2D CNN on 8x8 each timestep）
# 
class CNN_BiLSTM_TOF(nn.Module):
    """
    Input: (B, T, 5, 8, 8)
    Output: logits (B, 18)
    """
    def __init__(self, num_classes=18, emb_dim=128, hidden=128):
        super().__init__()
        self.cnn2d = nn.Sequential(
            nn.Conv2d(5, 32, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.Flatten(),                         # 64*8*8=4096
            nn.Linear(64 * 8 * 8, emb_dim),
            nn.ReLU(),
        )

        self.lstm = nn.LSTM(
            input_size=emb_dim,
            hidden_size=hidden,
            num_layers=1,
            batch_first=True,
            bidirectional=True
        )

        self.head = nn.Sequential(
            nn.Linear(hidden * 2, 128),
            nn.ReLU(),
            nn.Linear(128, num_classes)
        )

    def forward(self, x):
        # x: (B,T,5,8,8)
        B, T, C, H, W = x.shape
        x = x.reshape(B * T, C, H, W)      # (B*T, 5, 8, 8)
        emb = self.cnn2d(x)                # (B*T, emb_dim)
        emb = emb.reshape(B, T, -1)        # (B, T, emb_dim)

        out, _ = self.lstm(emb)            # (B, T, 2*hidden)
        last = out[:, -1, :]               # (B, 2*hidden)
        logits = self.head(last)           # (B, 18)
        return logits

model = CNN_BiLSTM_TOF(num_classes=18, emb_dim=128, hidden=128).to(device)
print(model)

# 
#    binary F1: sum(prob of 8 BFRB classes) >= 0.5
# 
def evaluate(model, loader, device, verbose=False):
    model.eval()
    criterion = nn.CrossEntropyLoss()

    total_loss = 0.0
    total_count = 0

    all_true_18, all_pred_18, all_logits = [], [], []

    with torch.no_grad():
        for x, y18, _ in loader:
            x = x.to(device)
            y18 = y18.to(device)

            logits = model(x)
            loss = criterion(logits, y18)

            bs = x.size(0)
            total_loss += loss.item() * bs
            total_count += bs

            pred = torch.argmax(logits, dim=1)

            all_true_18.append(y18.cpu().numpy())
            all_pred_18.append(pred.cpu().numpy())
            all_logits.append(logits.cpu().numpy())

    avg_loss = total_loss / max(total_count, 1)

    y_true_18 = np.concatenate(all_true_18).astype(int)
    y_pred_18 = np.concatenate(all_pred_18).astype(int)
    y_logits = np.concatenate(all_logits, axis=0)

    prob = torch.softmax(torch.tensor(y_logits), dim=1).numpy()

    p_bfrb = prob[:, BFRB_CLASS_INDICES].sum(axis=1)
    y_true_bin = np.isin(y_true_18, BFRB_CLASS_INDICES).astype(int)
    y_pred_bin = (p_bfrb >= 0.5).astype(int)
    f1_bin = f1_score(y_true_bin, y_pred_bin, average="binary", pos_label=1)

    # 9-class macro F1
    y_true_9 = convert_18_to_9_array(y_true_18)
    y_pred_9 = convert_18_to_9_array(y_pred_18)
    f1_macro9 = f1_score(y_true_9, y_pred_9, average="macro")

    if verbose:
        print(f"Eval | loss={avg_loss:.4f} | bin_f1={f1_bin:.4f} | macro_f1_9={f1_macro9:.4f}")

    return avg_loss, f1_bin, f1_macro9, y_true_18, y_pred_18, y_logits, prob
```

```python
### 3. TRAINING & EVALUATION STRATEGY
# - Optimizer: AdamW with a learning rate of 0.001 and weight decay of 1e-4[cite: 139].
# - Loss Function: CrossEntropyLoss for multi-class gesture classification.
# - Primary Metric (Binary F1): Calculated by summing probabilities of the 
#   8 BFRB classes and applying a 0.5 threshold[cite: 94, 141, 173].
# - Secondary Metric (Macro F1): Evaluates performance across 9 classes 
#   (8 BFRBs + 1 combined non-BFRB class)[cite: 176].
```

```python
LR = 1e-3
EPOCHS = 20

# 
optimizer = torch.optim.AdamW(model.parameters(), lr=LR, weight_decay=1e-4)
criterion = nn.CrossEntropyLoss()

train_losses, test_losses = [], []
test_f1_bin_list, test_f1_macro9_list = [], []

for epoch in range(1, EPOCHS + 1):
    model.train()
    running_loss = 0.0
    total = 0

    for x, y18, _ in train_loader:
        x = x.to(device)
        y18 = y18.to(device)

        optimizer.zero_grad()
        logits = model(x)
        loss = criterion(logits, y18)
        loss.backward()
        optimizer.step()

        bs = x.size(0)
        running_loss += loss.item() * bs
        total += bs

    train_loss = running_loss / max(total, 1)
    test_loss, f1_bin, f1_macro9, y_true_eval, y_pred_eval, _, _ = evaluate(model, test_loader, device)
    accuracy_val = (y_true_eval == y_pred_eval).mean()
    train_losses.append(train_loss)
    test_losses.append(test_loss)
    test_f1_bin_list.append(f1_bin)
    test_f1_macro9_list.append(f1_macro9)

    print(f"Epoch {epoch:02d}/{EPOCHS} | train_loss={train_loss:.4f} | test_loss={test_loss:.4f} | "
          f"binary_f1={f1_bin:.4f} | macro_f1(9c)={f1_macro9:.4f}")

# 
# Saved artifacts（late fusion）
#
ART_DIR = REPO_ROOT / "artifacts" / "cnn_bilstm_tof_seq"
MODEL_SUM = REPO_ROOT/ "results" /  "Model3_Training"
ART_DIR.mkdir(parents=True, exist_ok=True)
MODEL_SUM.mkdir(parents=True, exist_ok=True)

final_test_loss, final_f1_bin, final_f1_macro9, y_true_18, y_pred_18, y_logits, y_prob = evaluate(
    model, test_loader, device, verbose=True
)

np.save(ART_DIR / "y_true_18.npy", y_true_18)
np.save(ART_DIR / "y_pred_18.npy", y_pred_18)
np.save(ART_DIR / "y_logits.npy", y_logits) 
np.save(ART_DIR / "y_prob.npy", y_prob)

np.save(ART_DIR / "train_losses.npy", np.array(train_losses))
np.save(ART_DIR / "test_losses.npy", np.array(test_losses))
np.save(ART_DIR / "test_f1_bin.npy", np.array(test_f1_bin_list))
np.save(ART_DIR / "test_f1_macro9.npy", np.array(test_f1_macro9_list))

torch.save(model.state_dict(), ART_DIR / "model_weights.pt")
torch.save({
    "epoch": EPOCHS,
    "model_state": model.state_dict(),
    "optim_state": optimizer.state_dict(),
    "seed": SEED,
}, ART_DIR / "checkpoint.pt")

final_metrics = {
    "acc": float(accuracy_val),
    "macroF1": float(final_f1_macro9),
    "binaryF1": float(final_f1_bin),
}
with open(MODEL_SUM / "model3_training_summary.json", "w", encoding="utf-8") as f:
    json.dump(final_metrics, f, indent=2)

print("Saved to:", ART_DIR)
```

```python
import numpy as np
import torch
from pathlib import Path


PRED_DIR = REPO_ROOT / "results" / "preds"
PRED_DIR.mkdir(parents=True, exist_ok=True)

device = next(model.parameters()).device
model.eval()

all_logits = []
all_y = []

with torch.no_grad():
    for xb, yb, _ in test_loader:
        xb = xb.to(device)

        logits = model(xb)

        all_logits.append(logits.detach().cpu().numpy())
        all_y.append(yb.detach().cpu().numpy())

all_logits = np.concatenate(all_logits, axis=0)
all_y = np.concatenate(all_y, axis=0)

logits_path = PRED_DIR / "cnn_bilstm_tof_only_test_logits.npy"
y_path      = PRED_DIR / "cnn_bilstm_tof_only_test_y.npy"

np.save(logits_path, all_logits)
np.save(y_path, all_y)

print("Saved pred dir :", PRED_DIR.resolve())
print("Saved logits   :", logits_path.resolve(), "exists =", logits_path.exists())
print("Saved y        :", y_path.resolve(), "exists =", y_path.exists())
print("Logits shape   :", all_logits.shape, "Y shape:", all_y.shape)
```
